<h3 style="color:#1565C0;">
1. Introduction
</h3>


Food safety and public health are important concerns in urban areas. Regular health inspections of food establishments help government authorities identify potential health risks, monitor compliance with safety regulations, and ensure that businesses maintain appropriate hygiene standards.
       This project analyses the Health Inspection Scores dataset for the period 2016–2019. The dataset contains information about businesses and their health inspections, including inspection scores, types of inspections, violations, risk categories, business locations, and neighbourhood-related information.
 The purpose of this analysis is to explore patterns in health inspection results and identify important factors associated with food safety violations and inspection performance.


<h3 style="color:#1565C0;">
2. Problem Statement
</h3>

Food establishments may receive different health inspection scores depending on their compliance with health and safety regulations. Various factors, such as the type of inspection, risk category, geographical location, and the presence of violations, may influence the inspection outcomes.
The main problem addressed in this project is:
To analyse health inspection data and identify patterns, trends, and factors associated with inspection scores, violations, and food safety risks among businesses between 2016 and 2019.
The analysis will attempt to answer questions such as:
* How do health inspection scores vary among different businesses?
* What are the most common types of health and safety violations?
* How are violations distributed across different risk categories? 
* Do inspection scores vary according to inspection type?
* Are there geographical or neighbourhood-based patterns in health inspection results?
* How have inspection activities and scores changed over time?
* Which businesses or areas demonstrate a higher level of potential food safety risk?

Checking Outliers in Inspection Scores
</h3>

```python
Q1 = df_clean['inspection_score'].quantile(0.25)
Q3 = df_clean['inspection_score'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean['inspection_score'] < lower_bound) |
    (df_clean['inspection_score'] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

```python
print("Minimum inspection score:", df_clean['inspection_score'].min())
print("Maximum inspection score:", df_clean['inspection_score'].max())
```

Checking the Range of Inspection Scores:
The minimum and maximum inspection scores were calculated to identify the range and check for potentially invalid values. The minimum inspection score was 45, while the maximum score was 100. Since all recorded scores fall within the expected 0–100 range, there are no apparent invalid scores based on the score limits. Genuine low inspection scores will not be removed, as they may represent establishments with poorer inspection outcomes.

Handling Missing Inspection Scores:
The inspection_score column contains 14,432 missing values. Median imputation was considered; however, it was not applied because an inspection score represents an actual inspection outcome. Replacing missing scores with the median would create artificial observations and could alter the original distribution of scores and affect subsequent statistical analysis. Therefore, the missing values were retained as NaN. When analysis specifically requires inspection scores, records with missing scores will be excluded from that particular analysis while remaining available in the main cleaned dataset.

<h3 style="color:#1565C0;">10. Correlation Analysis</h3>

The correlation matrix reveals virtually no linear relationship between a business's health inspection score and its precise geographic coordinates, as shown by near-zero correlation values with latitude (0.01) and longitude (-0.01). This demonstrates that health compliance and food safety standards operate independently of a business's specific location within San Francisco. Meanwhile, the weak-to-moderate positive correlation of 0.26 between latitude and longitude simply reflects the natural diagonal geography of the city. Because raw coordinates lack predictive value for inspection scores, subsequent analyses should group data into categorical spatial features, such as postal codes or neighborhoods, to capture localized health safety trends.

<h3 style="color:#1565C0;">11. Inspection Scores by Risk Category and Inspection Type</h3>

A pivot table is used to examine average inspection scores across combinations of risk category and inspection type. This provides a multivariate view of inspection performance.

```python
pivot_table = pd.pivot_table(
    df_clean[df_clean['risk_category'] != 'Not Available'],
    values='inspection_score',
    index='risk_category',
    columns='inspection_type',
    aggfunc='mean',
)

risk_order = ['Low Risk', 'Moderate Risk', 'High Risk']
pivot_table = pivot_table.reindex(
    [r for r in risk_order if r in pivot_table.index]
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    pivot_table,
    annot=True,
    fmt='.1f',
    cmap='YlGnBu',
    linewidths=0.5,
    cbar_kws={'label': 'Mean Inspection Score'},
)

plt.title(
    'Average Inspection Score by Risk Category and Inspection Type',
    fontsize=14,
    pad=15,
)
plt.xlabel('Inspection Type', fontsize=12)
plt.ylabel('Risk Category', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

The pivot table excludes unscored administrative entries and unclassified risk categories (formerly **'Not Available'**) to focus strictly on actionable, scored routine inspections. Across these scored inspections, there is a clear, inverse relationship between the assigned risk level and the average inspection score: facilities categorized as **Low Risk** achieved the highest mean score at **87.1**, **Moderate Risk** followed at **85.4**, and **High Risk** recorded the lowest average score at **80.5**. This monotonic trend confirms that the risk classification system accurately reflects the operational hygiene and health compliance of inspected food establishments in San Francisco.
